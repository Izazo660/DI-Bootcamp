Exercise 1: Understanding Data Visualization

Why Data Visualization is Important

Data visualization translates complex raw data into a visual context (like charts, graphs, and maps). It is crucial in data analysis because:

Identifies Patterns and Trends: Human brains process visuals much faster than spreadsheets of raw numbers. Charts make anomalies, clusters, and trends instantly recognizable.

Simplifies Complexity: It condenses massive datasets into clear, digestible summaries.

Speeds Up Decision Making: Stakeholders can quickly grasp the insights derived from data, leading to faster, data-driven decisions.


Purpose of a Line Graph

The primary purpose of a line graph is to track changes over time (or across a continuous interval). By connecting individual data points with straight lines, it clearly illustrates whether data trends are increasing, decreasing, or fluctuating, making it ideal for metrics like daily temperatures, stock prices, or monthly revenue growth.

```python
#Exercise 2: Creating a Line Plot for Temperature Variation

import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temperatures = [72, 74, 76, 80, 82, 78, 75]

plt.figure(figsize=(8, 5))
plt.plot(days, temperatures, marker='o', color='crimson', linestyle='-', linewidth=2)

plt.title('Weekly Temperature Variation', fontsize=14, fontweight='bold')
plt.xlabel('Day', fontsize=12)
plt.ylabel('Temperature (°F)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

```python
#Exercise 3: Visualizing Monthly Sales with a Bar Chart

import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [5000, 5500, 6200, 7000, 7500]

plt.figure(figsize=(8, 5))
plt.bar(months, sales, color='royalblue', edgecolor='black', alpha=0.8)

plt.title('Monthly Sales Performance', fontsize=14, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Sales Amount ($)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
#Exercise 4: Visualizing the Distribution of CGPA

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Student_Mental_Health.csv')
df.columns = df.columns.str.strip()

df['What is your CGPA?'] = df['What is your CGPA?'].str.strip()

cgpa_order = ['0 - 1.99', '2.00 - 2.49', '2.50 - 2.99', '3.00 - 3.49', '3.50 - 4.00']
df['What is your CGPA?'] = pd.Categorical(df['What is your CGPA?'], categories=cgpa_order, ordered=True)

fig, ax = plt.subplots(figsize=(7, 4.5))
sns.histplot(data=df.sort_values('What is your CGPA?'), x='What is your CGPA?', ax=ax, color='teal', shrink=0.8)

ax.set_title("Distribution of Students' CGPA", fontsize=13, fontweight='bold')
ax.set_xlabel('CGPA Range', fontsize=11)
ax.set_ylabel('Count (Number of Students)', fontsize=11)
plt.xticks(rotation=15)

plt.tight_layout()
plt.show()
```

```python
#Exercise 5: Comparing Anxiety Levels Across Different Genders

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Student_Mental_Health.csv')
df.columns = df.columns.str.strip()

df['Anxiety_Numeric'] = df['Do you have Anxiety?'].map({'Yes': 1, 'No': 0})

anxiety_prop = df.groupby('Choose your gender')['Anxiety_Numeric'].mean().reset_index()
anxiety_prop = anxiety_prop.sort_values(by='Anxiety_Numeric')

fig, ax = plt.subplots(figsize=(6, 4.5))
sns.barplot(data=anxiety_prop, x='Choose your gender', y='Anxiety_Numeric', ax=ax, palette='muted')

ax.set_title('Proportion of Students Experiencing Anxiety by Gender', fontsize=13, fontweight='bold')
ax.set_xlabel('Gender', fontsize=11)
ax.set_ylabel('Proportion (Anxiety Rate)', fontsize=11)
ax.set_ylim(0, 1.0)

plt.tight_layout()
plt.show()
```

```python
#Exercise 6: Exploring the Relationship Between Age and Panic Attacks

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Student_Mental_Health.csv')
df.columns = df.columns.str.strip()

df['Panic_Attacks_Numeric'] = df['Do you have Panic attack?'].map({'Yes': 1, 'No': 0})

df_clean = df.dropna(subset=['Age'])

fig, ax = plt.subplots(figsize=(7, 4.5))
sns.scatterplot(
    data=df_clean, 
    x='Age', 
    y='Panic_Attacks_Numeric', 
    ax=ax, 
    color='purple', 
    marker='X', 
    s=120, 
    alpha=0.8
)

ax.set_title('Relationship Between Age and Panic Attacks', fontsize=13, fontweight='bold')
ax.set_xlabel('Age (Years)', fontsize=11)
ax.set_ylabel('Panic Attack Incident', fontsize=11)
ax.set_yticks([0, 1])
ax.set_yticklabels(['No (0)', 'Yes (1)'])

plt.tight_layout()
plt.show()
```
