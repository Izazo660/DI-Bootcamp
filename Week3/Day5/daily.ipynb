Daily Challenge: Interactive Data Visualization with Matplotlib and Seaborn

1. Data Preparation

Excel Serial Dates: The Order Date column is formatted as numeric values (Excel serial numbers) and has been successfully converted into standard YYYY-MM-DD format using origin='1899-12-30'.

Country Attribute: 100% of the observations are from the "United States", so distribution is broken down by state for a meaningful analysis.

2. Data Visualization with Matplotlib

Sales Trend: Sales slightly decreased from 2014 $484.2k to 2015 $470.5k, followed by strong exponential growth up to 2017 $733.2k.

Sales Distribution: Since the dataset contains only one country (US), sales distribution is represented using the top-performing US states (California and New York leading).

3. Data Visualization with Seaborn

Top 10 Products: The Canon imageCLASS 2200 Advanced Copier is by far the highest revenue-generating product with over $61,500 in total sales.

Profit vs. Discount: The scatter plot clearly shows that high discounts (above 20%-30%) heavily correspond to negative profits (losses).

4. Comparative Analysis

Matplotlib: Highly structured and granular, ideal for core low-level adjustments (e.g., specific line markers and axis rotations).

Seaborn: Extremely easy to use with high-level functions, automatically applies clean color palettes (viridis), and handles grouped categoricals efficiently with minimal syntax.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 1. DATA PREPARATION
df = pd.read_csv("US_Superstore_data.xls - Orders.csv")

df['Order_Date_Clean'] = pd.to_datetime(df['Order Date'], unit='D', origin='1899-12-30')
df['Year'] = df['Order_Date_Clean'].dt.year


# 2. DATA VISUALIZATION WITH MATPLOTLIB

yearly_sales = df.groupby('Year')['Sales'].sum().reset_index()
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(yearly_sales['Year'].astype(str), yearly_sales['Sales'], marker='o', color='b', linewidth=2)
ax.set_title('Sales Trends Over the Years (2014-2017)')
ax.set_xlabel('Year')
ax.set_ylabel('Total Sales ($)')
plt.tight_layout()
plt.savefig('matplotlib_yearly_sales.png')
plt.close()

state_sales = df.groupby('State')['Sales'].sum().reset_index().sort_values(by='Sales', ascending=False)
fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(state_sales['State'].head(10), state_sales['Sales'].head(10), color='teal')
ax.set_title('Top 10 US States by Sales Distribution')
ax.set_xlabel('State')
ax.set_ylabel('Total Sales ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('matplotlib_sales_distribution.png')
plt.close()


# 3. DATA VISUALIZATION WITH SEABORN

top_products = df.groupby('Product Name')['Sales'].sum().reset_index().sort_values(by='Sales', ascending=False).head(10)
fig, ax = plt.subplots(figsize=(8, 5))
sns.barplot(data=top_products, x='Sales', y='Product Name', ax=ax, palette='viridis', hue='Product Name')
if ax.get_legend() is not None:
    ax.get_legend().remove()
ax.set_title('Top 10 Products by Total Sales')
ax.set_xlabel('Total Sales ($)')
ax.set_ylabel('Product Name')
plt.tight_layout()
plt.savefig('seaborn_top_products.png')
plt.close()

fig, ax = plt.subplots(figsize=(7, 4))
sns.scatterplot(data=df, x='Discount', y='Profit', ax=ax, alpha=0.6, color='coral')
ax.set_title('Relationship between Profit and Discount')
ax.set_xlabel('Discount Level')
ax.set_ylabel('Profit / Loss ($)')
plt.tight_layout()
plt.savefig('seaborn_profit_vs_discount.png')
plt.close()

print("All tasks executed and visualizations saved to your workspace.")
```
