```python
#Exercise 2: Descriptive Statistics

import numpy as np
from scipy import stats

data = [12, 15, 13, 12, 18, 20, 22, 21]

mean = np.mean(data)
median = np.median(data)
variance = stats.tvar(data) 
std_dev = stats.tstd(data)

print(f"mean : {mean}")
print(f"median : {median}")
print(f"variance : {variance:.2f}")
print(f"standard Deviation : {std_dev:.2f}")
```

```python
#Exercise 3: Understanding Distributions

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mean = 50
std_dev = 10

x = np.linspace(mean - 4*std_dev, mean + 4*std_dev, 100)

y = norm.pdf(x, mean, std_dev)

plt.plot(x, y, label=f'Normal Dist (μ={mean}, σ={std_dev})', color='blue')
plt.title('Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Exercise 4: T-Test Application

import numpy as np
from scipy import stats

np.random.seed(42)
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

t_stat, p_value = stats.ttest_ind(data1, data2)

print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4e}")

alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis : The means are significantly different.")
else:
    print("Fail to reject the null hypothesis : No significant difference between means.")
```

```python
#Exercise 5: Linear Regression Analysis

import numpy as np
from scipy import stats

house_sizes = np.array([50, 70, 80, 100, 120])
house_prices = np.array([150000, 200000, 210000, 250000, 280000])

slope, intercept, r_value, p_value, std_err = stats.linregress(house_sizes, house_prices)

print(f"Slope : {slope:.2f}")
print(f"Intercept : {intercept:.2f}")

target_size = 90
predicted_price = slope * target_size + intercept
print(f"Predicted price for a 90 sqm house : {predicted_price:,.2f} currency units")
```

```python
#Exercise 6: Understanding ANOVA

from scipy import stats

fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

f_stat, p_value = stats.f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)

print(f"F-value: {f_stat:.4f}")
print(f"P-value: {p_value:.5f}")
```

F-value and P-value: F-value $\approx 28.38$, P-value $\approx 0.00003$.

Do they have significantly different effects ? 

Yes. Because the P-value ($0.00003$) is substantially lower than the standard significance level ($\alpha = 0.05$), we reject the null hypothesis. There is strong statistical evidence that at least one fertilizer produces different plant growth than the others.

What if the P-value were greater than 0.05?

If $P > 0.05$, we would fail to reject the null hypothesis. This would imply that any observed differences in plant growth between the three groups could easily be attributed to random chance, meaning we couldn't conclude that the fertilizers have different effects.
