```python
#Task 1: Data Import & Cleaning

import pandas as pd
import numpy as np

df = pd.read_csv("Airplane_Crashes_and_Fatalities_Since_1908_t0_2023.csv")

df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df['Year'] = df['Date'].dt.year
df['Decade'] = (df['Year'] // 10) * 10

df['Fatalities'] = pd.to_numeric(df['Fatalities'], errors='coerce').fillna(0)
df['Aboard'] = pd.to_numeric(df['Aboard'], errors='coerce').fillna(0)

df['Survival_Rate'] = np.where(df['Aboard'] > 0, (df['Aboard'] - df['Fatalities']) / df['Aboard'], 0.0)
df['Survival_Rate'] = df['Survival_Rate'].clip(0.0, 1.0)

print(f"Dataset successfully cleaned. Total records : {len(df)}")
```

```python
#Task 2: Exploratory Data Analysis (EDA)

total_crashes = len(df)
total_fatalities = df['Fatalities'].sum()
avg_survival_rate = df['Survival_Rate'].mean() * 100

print("GLOBAL AVIATION CRASH METRICS")
print(f"Total Documented Crashes : {total_crashes}")
print(f"Total Historical Fatalities : {total_fatalities:,.0f}")
print(f"Average Survival Rate per Incident : {avg_survival_rate:.2f}%")
```

```python
#Task 3: Statistical Analysis (SciPy Application)

from scipy import stats

print("FATALITIES DISTRIBUTION STATS")
print(f"Mean Fatalities  : {df['Fatalities'].mean():.2f}")
print(f"Median Fatalities: {df['Fatalities'].median():.2f}")
print(f"Std Deviation    : {df['Fatalities'].std():.2f}")
```

```python
fatalities_1970s = df[df['Decade'] == 1970]['Fatalities']
fatalities_2010s = df[df['Decade'] == 2010]['Fatalities']

t_stat, p_val = stats.ttest_ind(fatalities_1970s, fatalities_2010s, equal_var=False)

print("\nHYPOTHESIS TEST: 1970s vs 2010s FATALITIES")
print(f"1970s Mean Fatalities per crash : {fatalities_1970s.mean():.2f}")
print(f"2010s Mean Fatalities per crash : {fatalities_2010s.mean():.2f}")
print(f"T-statistic : {t_stat:.4f}")
print(f"P-value : {p_val:.5e}")

alpha = 0.05
if p_val < alpha:
    print("Result: Reject H0. There is a statistically significant difference in fatalities between eras.")
else:
    print("Result: Fail to reject H0. No statistically significant difference found.")
```

```python
#Task 4: Visualization Block

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="darkgrid")
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

yearly_crashes = df.groupby('Year').size()
axes[0, 0].plot(yearly_crashes.index, yearly_crashes.values, color='crimson', linewidth=2)
axes[0, 0].set_title('Global Trend of Airplane Crashes over Years')
axes[0, 0].set_xlabel('Year')
axes[0, 0].set_ylabel('Number of Accidents')

sns.histplot(df['Fatalities'], bins=40, ax=axes[0, 1], color='darkblue', kde=True)
axes[0, 1].set_yscale('log')
axes[0, 1].set_title('Distribution of Fatalities per Incident (Log Scale)')
axes[0, 1].set_xlabel('Fatalities Count')

top_operators = df['Operator'].value_counts().head(10)
sns.barplot(x=top_operators.values, y=top_operators.index, ax=axes[1, 0], palette='viridis')
axes[1, 0].set_title('Top 10 Operators by Incident Count')
axes[1, 0].set_xlabel('Total Crashes')

decades_compare = df[df['Decade'].isin([1970, 2010])]
sns.boxplot(x='Decade', y='Fatalities', data=decades_compare, ax=axes[1, 1], palette='Set2')
axes[1, 1].set_title('Fatalities per Crash: 1970s vs 2010s')
axes[1, 1].set_ylim(-5, 150)

plt.tight_layout()
plt.show()
```

Task 5: Summary & Insights Report

1. The Incident Curve Peak (Exploratory Findings)

The time-series evaluation reveals that aviation accidents climbed steadily through the early 20th century, peaking heavily between the 1940s and late 1970s. This spike correlates with the mass adoption of military aviation around WWII followed by the rapid expansion of early commercial jet travel before modern fly-by-wire and computer-assisted safety protocols were standardized. Post-1990s, the crash trend shows a steady, steep decline.

2. High Skewness in Fatalities (Statistical Distribution)

The distribution of fatalities per incident shows a heavy right-skew. The median number of fatalities per crash is surprisingly low compared to the mean. This tells us that a massive portion of historical records consists of light aircraft, cargo flights, or minor incidents with very few casualties, while massive commercial aircraft disasters form a thin, high-impact tail on the right side of the distribution.

3. Hypothesis Test Evaluation

Our independent $T$-test between the 1970s and 2010s yielded a P-value significantly below our $\alpha = 0.05$ threshold ($P < 0.001$).

The Paradigm Shift: Despite modern commercial flights carrying significantly more passengers simultaneously, the average fatalities per crash dropping down significantly in the 2010s proves a statistically measurable leap forward in aviation safety engineering, crashworthiness, and faster emergency response times.
