```python
#Daily Challenge: Data Handling and Analysis in Python

import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA

df = pd.read_csv('datascience_salaries.csv')

print(df.groupby('experience_level')['salary'].agg(['mean', 'median']))

scaler = MinMaxScaler()
df['salary_normalized'] = scaler.fit_transform(df[['salary']])

df_numeric = pd.get_dummies(df, drop_first=True)
df_numeric = df_numeric.select_dtypes(include=['number'])

pca = PCA(n_components=2)
pca_results = pca.fit_transform(df_numeric)

df['pca_1'] = pca_results[:, 0]
df['pca_2'] = pca_results[:, 1]

print(df[['pca_1', 'pca_2', 'salary_normalized']].head())
```
