Exercise 1: Introduction to Data Analysis

Today, we are surrounded by data. Every time we click a link, buy something online, or use a phone app, we create digital information. But raw data by itself is just a giant, messy pile of numbers. Data analysis is the process of cleaning, organizing, and studying that information to find meaningful patterns and make smart decisions. It usually follows a simple journey. First, you scrub the data to fix mistakes and remove duplicates. Next, you use software or formulas to find trends. And finally, you turn those results into clear charts so that anyone can understand them at a glance.

In the modern world, companies and organizations cannot afford to rely on gut feelings or luck anymore. Data analysis is incredibly important because it replaces guesswork with solid proof. It helps businesses understand what their customers actually want, shows them how to stop wasting money, and allows them to predict future market shifts instead of just reacting to them when it is too late. Basically, it gives organizations a clear map to navigate a very competitive digital landscape.

You can see this tool at work across almost every industry today. In healthcare, hospitals analyze patient records and real-time vitals to predict dangerous conditions hours before they physically show up. In e-commerce, platforms like Amazon or Netflix track your browsing history to recommend products tailored exactly to your taste, while automatically adjusting prices based on current demand. Even in sports, professional teams use wearable sensors and cameras to track player movements, analyzing the data to build better game tactics and prevent injuries. Ultimately, data analysis is all about turning chaotic information into a clear plan to solve real-world problems.

Exercise 5: Basic Observation Skills in Data Analysis

Trend Analysis

Columns: Year and Avg hrs per day sleeping

Explanation: By tracking Avg hrs per day sleeping across the chronological timeline of Year, you can determine whether Americans are gaining or losing sleep over time, identifying long-term public health behaviors.

Group Comparison Analysis

Columns: Sex, Age Group, and Avg hrs per day sleeping

Explanation: This allows you to compare average sleep durations between different demographic groups (e.g., Men vs. Women, or Young Adults vs. Seniors) to pinpoint which segments of the population are most sleep-deprived.

Contextual / Day-Type Analysis

Columns: Type of Days (or Period) and Avg hrs per day sleeping

Explanation: This helps analyze the difference in sleeping habits on weekdays versus weekends/holidays, revealing how work-life schedules impact rest patterns.

Exercise 7: Transformation Exercise

1. A series of blog posts about travel experiences

Proposed Method: Apply Natural Language Processing (NLP), specifically Named Entity Recognition (NER) and Sentiment Analysis.

Resulting Structure: A tabular database (Excel/SQL) with columns like: Post_ID, Destination_Country, City, Travel_Duration_Days, Key_Activities, and Sentiment_Score (positive/negative experience rating).

Reasoning: Blog posts contain free-form storytelling. By extracting specific entities (places, times) and assigning a quantitative value to the mood (sentiment), you transform essays into searchable, filterable data rows.

2. Audio recordings of customer service calls

Proposed Method: Implement a two-step pipeline using Speech-to-Text Transcription models followed by Text Mining/Classification algorithms.

Resulting Structure: A structured table with columns like: Call_ID, Timestamp, Call_Duration, Customer_Issue_Category (e.g., billing, technical support), Resolution_Status (Yes/No), and Agent_ID.

Reasoning: Audio files cannot be easily searched or queried directly. Converting them to digital text allows machine learning models to instantly categorize the main reason for the call and quantify customer satisfaction metrics.

3. Handwritten notes from a brainstorming session

Proposed Method: Use Intelligent Character Recognition (ICR) or Optical Character Recognition (OCR) software tailored for handwriting, followed by tag/keyword grouping.

Resulting Structure: A structured spreadsheet or Kanban dataset with columns like: Idea_ID, Core_Theme, Suggested_Action_Item, Assigned_To, and Priority_Level.

Reasoning: Handwritten notes are highly chaotic and physically confined. Digitizing the text with OCR and sorting them by functional tags transforms casual sketches into an actionable project management database.

4. A video tutorial on cooking

Proposed Method: Utilize Video-to-Text Translation (extracting the voice track transcript) combined with Computer Vision (Object Detection) to detect visual milestones.

Resulting Structure: A highly structured recipe profile consisting of two tables:

    Ingredients Table: Ingredient_Name, Quantity, Unit.

    Steps Table: Step_Number, Action/Technique (e.g., chop, boil), Target_Temperature, Duration_Minutes.

Reasoning: A video is a dense stream of continuous audio-visual information. Breaking it down into timestamps, distinct ingredients, and numerical steps turns a visual guide into an organized relational dataset.

```python
#Exercise 9 : Export a dataframe to excel format and JSON format.

import pandas as pd

data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 32],
    'City': ['New York', 'Paris', 'Berlin', 'London']
}
df = pd.DataFrame(data)
print("--- Created DataFrame ---")
print(df)

df.to_excel('simple_dataframe.xlsx', index=False)
print("\nDataFrame successfully exported to 'simple_dataframe.xlsx'")


df.to_json('simple_dataframe.json', orient='records', indent=4)
print("DataFrame successfully exported to 'simple_dataframe.json'")
```
