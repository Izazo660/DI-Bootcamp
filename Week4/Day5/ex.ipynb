Exercise 1: Defining the Problem and Data Collection for Loan Default Prediction

The objective of this project is to predict the target variable, which is whether an applicant will default on their loan, to support the financial institution's decision to approve or reject a credit application.
To achieve this, I would collect demographic data such as age and employment length. Employment history helps evaluate the applicant's overall financial stability. I would also gather financial data like annual income and debt-to-income ratios. These metrics assess the borrower's current capacity to absorb and repay new debt. Next, I would collect credit history, including credit scores and past delinquencies. This behavioral data is crucial because past financial responsibility is a strong predictor of future repayment. Finally, I would retrieve loan details, such as the loan amount and interest rate. The requested amount contextualizes the financial burden this new loan will place on the borrower.

The data would originate from the financial institution's internal records, extracted from their SQL databases to analyze existing customer history. For new clients or profile enrichment, data would be collected from credit bureaus (like Equifax or Experian) using secure API integrations to pull real-time credit histories.

This project involves significant risks and constraints regarding privacy and regulation, as it handles highly sensitive personal financial data subject to strict legal frameworks. There is also a major risk of sampling bias if the model is trained exclusively on historically approved loans, ignoring those that were rejected. Furthermore, data quality and governance must be tightly controlled to prevent missing values or data corruption from causing discriminatory or unfair lending decisions.

Exercise 2: Feature Selection and Model Choice for Loan Default Prediction

The features I believe are most relevant are credit_score, debt_to_income, annual_income, loan_amount, and num_delinquencies.
The credit score mathematically summarizes an individual's past financial reliability, which correlates heavily with default risk. The debt-to-income ratio and annual income demonstrate the applicant's mathematical capacity to handle monthly payments. The loan amount is relevant because larger debts increase the financial strain on the borrower. Past delinquencies serve as a direct behavioral signal that the individual has struggled to meet financial obligations before.
I would exclude the zip_code feature because using geographic location can lead to ethical bias or redlining, potentially discriminating against certain neighborhoods. I would also exclude age if local regulations strictly prohibit its use, ensuring compliance with fair lending laws.

To handle categorical variables, I would use one-hot encoding for nominal features like loan purpose, and ordinal encoding for features with an inherent hierarchy like employment length. To impute missing values, I would use the median for numerical features to avoid the influence of outliers, and the mode for categorical variables.

Exercise 3: Training, Evaluating, and Optimizing the Model

For this classification problem, I would choose a Random Forest or an XGBoost model because tree-based ensembles perform exceptionally well on tabular financial data and inherently capture complex non-linear relationships.
My evaluation plan will split the data into training (80%) and testing (20%) sets, utilizing a 5-fold cross-validation strategy on the training set. I will report Precision, Recall, F1-score, and ROC-AUC metrics, choosing a decision threshold that minimizes costly false negatives (undetected defaults).
To address the severe class imbalance typical of loan datasets, I will use stratified splitting and apply class weights within the algorithm to penalize errors on the minority class more heavily.
To improve performance while strictly avoiding data leakage, I will iterate on hyperparameters using Grid Search exclusively within the cross-validation loops of the training set. The final optimized model will only be evaluated on the test set once at the very end of the pipeline.

Exercise 4: Designing Machine Learning Solutions for Specific Problems

Predicting Stock Prices
This scenario requires Supervised Learning (specifically regression or time-series forecasting) because we utilize labeled historical data to predict a continuous numerical value. The input data consists of historical stock prices, trading volumes, and market indicators over time. The output is a continuous numerical value representing the estimated future stock price. The learning signal aims to minimize the error (such as Mean Squared Error) between the predicted price and the actual market price that occurs.

Organizing a Library of Books
This problem requires Unsupervised Learning (specifically clustering) because the books do not have predefined category labels and must be grouped based on inherent structural similarities. The input data includes book metadata, text summaries, and keywords. The output is a specific cluster assignment for each book, grouping similar literature together. The learning objective is to maximize similarity within each cluster while maximizing the distance between different clusters.

Program a robot to navigate and find the shortest path in a maze
This scenario requires Reinforcement Learning because an agent must learn the optimal sequence of actions through trial and error by interacting with a dynamic environment. The input data is the robot's current state, including its position, orientation, and sensor readings of nearby walls. The output is a discrete action taken by the robot, such as moving forward, turning left, or turning right. The learning signal is a cumulative reward that penalizes colliding with walls or wasting time and rewards reaching the exit quickly.

Exercise 5: Designing an Evaluation Strategy for Different ML Models

For a supervised classification model, the evaluation strategy relies on a train-test split combined with k-fold cross-validation to ensure the model generalizes well to unseen data. I will report accuracy, precision, recall, F1-score, and evaluate the model using ROC and Precision-Recall curves to assess performance across different thresholds. A key evaluation challenge in this category is selecting the optimal decision threshold when the business costs of false positives and false negatives are highly asymmetrical.

To assess an unsupervised clustering model, I would measure the mathematical quality and cohesion of the generated groups using the silhouette score. I would also use diagnostic tools such as PCA plots for visual cluster separation and the elbow method to determine the optimal number of clusters. A key evaluation challenge here is the complete lack of ground truth labels, making it difficult to verify if the clusters represent meaningful real-world categories.

For a reinforcement learning model, I would track the average cumulative reward per episode over time to ensure the agent is successfully learning an optimal policy. I would evaluate convergence by monitoring when the reward curve flattens out, while using an epsilon-greedy decay strategy to carefully balance exploration of new paths and exploitation of known rewards. A major evaluation challenge in reinforcement learning is designing a flawless reward function, as agents often exploit loopholes to maximize points without actually solving the intended problem.
