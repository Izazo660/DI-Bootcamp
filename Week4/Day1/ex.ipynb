```python
import numpy as np
```

Exercise 1: Array Creation and Manipulation

```python
arr1 = np.arange(10)
print(arr1)
```

Exercise 2: Type Conversion and Array Operations

```python
lst = [3.14, 2.17, 0, 1, 2]
arr2 = np.array(lst).astype(int)
print(arr2)
```

Exercise 3: Working with Multi-Dimensional Arrays

```python
arr3 = np.arange(1, 10).reshape(3, 3)
print(arr3)
```

Exercise 4: Creating Multi-Dimensional Array with Random Numbers

```python
arr4 = np.random.rand(4, 5)
print(arr4)
```

Exercise 5: Indexing Arrays

```python
array_2d = np.array([[21, 22, 23, 22, 22], [20, 21, 22, 23, 24], [21, 22, 23, 22, 22]])

second_row = array_2d[1]
print(second_row)
```

Exercise 6: Reversing elements

```python
arr6 = np.arange(10)
reversed_arr6 = arr6[::-1]
print(reversed_arr6)
```

Exercise 7: Identity Matrix

```python
identity_matrix = np.eye(4)
print(identity_matrix)
```

Exercise 8: Simple Aggregate Funcs

```python
arr8 = np.arange(10)

arr_sum = np.sum(arr8)
arr_avg = np.mean(arr8)

print(f"Sum: {arr_sum}, Average: {arr_avg}")
```

Exercise 9: Create Array and Change its Structure

```python
arr9 = np.arange(1, 21).reshape(4, 5)
print(arr9)
```

Exercise 10: Conditional Selection of Values

```python
arr10 = np.arange(10)

odd_numbers = arr10[arr10 % 2 != 0]
print(odd_numbers)
```
