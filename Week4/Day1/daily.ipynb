Daily Challenge: NumPy, Pandas, and Matplotlib Integration

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

np.random.seed(42)

# 1. Data Preparation

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
cities = [f"City {i}" for i in range(1, 11)]

temp_data = np.random.uniform(low=-5, high=35, size=(10, 12))

df_temp = pd.DataFrame(data=temp_data, index=cities, columns=months)

# 2. Data Analysis

df_temp["Annual_Avg"] = df_temp.mean(axis=1)

highest_city = df_temp["Annual_Avg"].idxmax()
lowest_city = df_temp["Annual_Avg"].idxmin()

print("--- Temperature Analysis ---")
print(df_temp[["Annual_Avg"]].round(2))
print(f"\nCity with highest average temperature: {highest_city} ({df_temp.loc[highest_city, 'Annual_Avg']:.2f}°C)")
print(f"City with lowest average temperature: {lowest_city} ({df_temp.loc[lowest_city, 'Annual_Avg']:.2f}°C)\n")
df_plot = df_temp.drop(columns=["Annual_Avg"])

# 3. Data Visualization

fig, ax = plt.subplots(figsize=(12, 6))

colors = [
    "#46b59b",
    "#61b392",
    "#2d7866",
    "#59ccb0",
    "#4a967c",
    "#236355",
    "#00a877",
    "#3f9179",
    "#437367",
    "#02a877",
]

for i, city in enumerate(df_plot.index):
    if city in ["City 7", "City 10"]:
        linestyle = "--"
        linewidth = 2.5
    else:
        linestyle = "-"
        linewidth = 1.3

    ax.plot(
        df_plot.columns,
        df_plot.loc[city],
        label=city,
        color=colors[i],
        linestyle=linestyle,
        linewidth=linewidth,
    )

ax.set_title("Monthly Temperature Trends for Each City", fontsize=16, pad=10, color="#222222")
ax.set_xlabel("Month", fontsize=13, labelpad=5)
ax.set_ylabel("Temperature (°C)", fontsize=13, labelpad=5)

ax.grid(True, linestyle="--", alpha=0.5, color="#bcbcbc")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_color("#222222")
ax.spines["bottom"].set_color("#222222")

ax.tick_params(axis="both", labelsize=12, colors="#222222")

ax.legend(loc="lower left", frameon=True, facecolor="white", edgecolor="#e0e0e0")

plt.tight_layout()
plt.show()
```
