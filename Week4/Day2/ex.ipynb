Exercise 1: Matrix Operations

```python
matrix = np.array([[2, 1, 3], 
                   [0, 5, 6], 
                   [1, 0, 2]])

det = np.linalg.det(matrix)

if det != 0:
    inverse = np.linalg.inv(matrix)
else:
    inverse = "Matrix is singular and cannot be inverted."

print("Original Matrix:\n", matrix)
print("\nDeterminant:", round(det, 4))
print("\nInverse Matrix:\n", inverse)
```

Exercise 6: Basic Hypothesis Testing

```python
np.random.seed(42)
productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Hypothesis Formulation
# Null Hypothesis (H0): The training program had no effect. The mean difference is <= 0.
# Alternative Hypothesis (H1): The training program increased productivity. The mean difference > 0.


differences = productivity_after - productivity_before

mean_improvement = np.mean(differences)
std_error = np.std(differences, ddof=1) / np.sqrt(len(differences))

t_stat = mean_improvement / std_error

print(f"Mean Productivity Improvement: {mean_improvement:.2f}")
print(f"Calculated T-Statistic: {t_stat:.2f}")

# Conclusion rule-of-thumb: A t-statistic greater than 2.04 (for 29 degrees of freedom at a 95% confidence level) 
# implies we reject the null hypothesis.
if t_stat > 2.04:
    print("Conclusion: Reject H0. The training program significantly improved productivity!")
else:
    print("Conclusion: Fail to reject H0. No significant productivity improvement detected.")
```

Exercise 8: Time Series Data Manipulation

Exercise 10: Basic Visualization

```python
np.random.seed(10)
time_steps = np.arange(1, 101)
data_points = np.random.randn(100).cumsum()

plt.figure(figsize=(10, 5))
plt.plot(time_steps, data_points, color='teal', linewidth=2, label='Random Walk')

plt.title("Basic Visualization: NumPy Data on Matplotlib Line Graph", fontsize=14)
plt.xlabel("Time Steps", fontsize=11)
plt.ylabel("Value", fontsize=11)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.show()
```
